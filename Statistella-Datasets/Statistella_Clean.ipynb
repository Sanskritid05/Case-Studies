```python
# dealing with nans, not converting it to 0
stat_cols = [
    'PTS_home', 'FG_PCT_home', 'FT_PCT_home', 'FG3_PCT_home',
    'AST_home', 'REB_home',
    'PTS_away', 'FG_PCT_away', 'FT_PCT_away', 'FG3_PCT_away',
    'AST_away', 'REB_away'
]
```

```python
final_games = games_data[
    games_data['GAME_STATUS_TEXT'] == 'Final'
].copy()
```

```python
final_games[stat_cols].isnull().sum()
```

```python
final_games[stat_cols] = final_games[stat_cols].apply(
    pd.to_numeric, errors='coerce'
)
```
